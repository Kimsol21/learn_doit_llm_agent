#### 3. 기본 에이전트 구축

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """지정된 도시의 날씨를 가져옵니다."""
    return f"{city}의 날씨는 맑습니다!"

agent = create_agent(
    model="openai:gpt-5.5",
    tools=[get_weather],
    system_prompt="너는 날씨 정보를 제공하는 도우미야. 사용자가 도시 이름을 입력하면 해당 도시의 날씨를 알려줘.",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "런던의 날씨는 어떻게 되나요?"}]}
)
print(result["messages"][-1].content_blocks)
```

- 에이전트는 사용자의 질문을 이해하고, 도시 이름("런던")을 인자로 삼아 날씨 도구(`get_weather`)를 자동으로 호출.

```python
from rich.pretty import pprint

pprint(result)
```

#### 4. 실전 에이전트 구축
- 텍스트 파일에 대한 질문에 답하는 연구용 에이전트 구축

##### 시스템 프롬프트 정의

```python
SYSTEM_PROMPT = """You are a literary data assistant.

## Capabilities

- `fetch_text_from_url`: loads document text from a URL into the conversation.
Do not guess line counts or positions—ground them in tool results from the saved file."""
```

#### @tool 데코레이터를 사용하여 URL에서 문서 텍스트를 불러오는 fetch_text_from_url 도구를 정의.

```python
import urllib.error
import urllib.request

from langchain.tools import tool


@tool
def fetch_text_from_url(url: str) -> str:
    """Fetch the document from a URL.
    """
    req = urllib.request.Request(
        url,
        headers={"User-Agent": "Mozilla/5.0 (compatible; quickstart-research/1.0)"},
    )
    try:
        with urllib.request.urlopen(req, timeout=120) as resp:
            raw = resp.read()
    except urllib.error.URLError as e:
        return f"Fetch failed: {e}"
    text = raw.decode("utf-8", errors="replace")
    return text
```

#### 모델 구성
- `init_chat_model`을 사용하여 온도(`temperature`), 타임아웃, 최대 토큰(`max_tokens`) 등을 설정

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "openai:gpt-5.5",
    temperature=0.5,
    timeout=300,
    max_tokens=25000,
)
```

#### 대화 상태 및 상호작용 이력을 유지하기 위해 InMemorySaver 체크포인터를 추가.

```python
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()
```

#### 에이전트 생성 및 실행
-  **LangChain Agents**: 세밀한 제어가 필요할 때 선택.
-  **Deep Agents (** **create\_deep\_agent** **)**: 계획 수립, 파일 시스템 접근, 서브에이전트 생성 기능이 기본 내장되어 있어 복잡한 작업에 적합 (`uv add deepagents` 설치 필요)

```python
from langchain.agents import create_agent
from deepagents import create_deep_agent

agent = create_agent(
    model=model,
    tools=[fetch_text_from_url],
    system_prompt=SYSTEM_PROMPT,
    checkpointer=checkpointer,
)

deep_agent = create_deep_agent(
    model=model,
    tools=[fetch_text_from_url],
    system_prompt=SYSTEM_PROMPT,
    checkpointer=checkpointer,
)

content = f"""Project Gutenberg에서 F. 스콧 피츠제럴드의 《위대한 개츠비》 전체 평문 파일을 제공합니다.
URL: https://www.gutenberg.org/files/64317/64317-0.txt

가능한 한 다음 질문에 답변하세요.

1) 전체 Gutenberg 파일에서 `Gatsby`라는 부분 문자열을 포함하는 줄은 몇 줄인가요? (줄 안의 출현 횟수가 아니라 줄의 개수를 세며, 각 줄은 줄바꿈 문자로 끝난다고 가정합니다.)
2) 파일에서 `Daisy`를 포함하는 첫 번째 줄의 1부터 시작하는 줄 번호는 몇 번인가요?
3) 중립적인 요약을 두 문장으로 작성하세요.

(1)과 (2)는 최선을 다해 답변하세요. 사용 가능한 도구와 추론만으로 정확한 답을 **검증**할 수 없다는 것을 알게 되면 숫자를 지어내지 말고, 해당 필드에 `null`을 사용하세요. 또한 `how_you_computed_counts`에 그 한계를 명시하세요. 답변은 한국어로 작성하세요. 오류가 발생하면 오류의 내용과 오류 메시지를 함께 보고하세요."""

print("create_agent 실행 중...", flush=True)
agent_result = agent.invoke(
    {"messages": [{"role": "user", "content": content}]},
    config={"configurable": {"thread_id": "great-gatsby-lc"}},
)
print("create_deep_agent 실행 중...", flush=True)
deep_agent_result = deep_agent.invoke(
    {"messages": [{"role": "user", "content": content}]},
    config={"configurable": {"thread_id": "great-gatsby-da"}},
)
print("\ncreate_agent 결과:")
print(agent_result["messages"][-1].content_blocks)
print("\ncreate_deep_agent 결과:")
print(deep_agent_result["messages"][-1].content_blocks)
```

### Deep Agent와 LangChain 에이전트 비교

| 구분 | LangChain 에이전트 | Deep Agent |
|---|---|---|
| 기본 기능 | 모델과 도구 중심의 기본 에이전트 | 계획, 파일 시스템, 서브에이전트 기능 내장 |
| 복잡한 작업 | 필요한 기능을 직접 구현해야 함 | `write_todos`로 작업을 계획하고 단계별 수행 |
| 컨텍스트 관리 | 도구를 추가해 직접 구성 | `grep`, `read_file` 등으로 자동 관리 |
| 적합한 용도 | 단순하고 세밀한 제어가 필요한 작업 | 장시간 수행되는 복잡한 연구 및 분석 작업 |

Deep Agent는 LangChain 에이전트보다 복잡한 작업을 위한 기능이 풍부하며, LangChain 에이전트는 필요한 기능을 직접 조합해 유연하게 구성할 수 있다.

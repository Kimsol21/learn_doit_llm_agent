### Ollama 란?
- 사용자의 로컬 컴퓨터 환경에서 대규모 언어 모델(LLM)을 손쉽게 다운로드하고 실행할 수 있게 해주는 무료 오픈소스 플랫폼입니다. 
- 복잡한 설정 없이 명령어 한 줄로 ChatGPT 같은 AI를 내 PC에서 직접 구동할 수 있습니다.
#### 주요 특징
- 개인정보 보호 : 인터넷 연결 없이 오프라인에서도 작동하므로 대화 내용이나 민감한 데이터가 외부 서버로 유출되지 않습니다.
- 비용 절감 : 유료 API 키나 구독료 없이 무제한으로 AI 모델을 사용할 수 있습니다.
- 다양한 모델 지원: Llama, Qwen, Mistral, Gemma 등 성능이 검증된 다양한 오픈소스 AI 모델을 지원합니다.
- 쉬운 사용법: 도커(Docker)와 유사한 방식으로 명령어를 통해 모델을 관리할 수 있습니다
#### 사용법
| 명령어 | 설명 |
|---|---|
| `ollama run <model>` | 모델 실행 |
| `ollama pull <model>` | 모델 다운로드 |
| `ollama list` | 다운로드한 모델 확인 |
| `ollama show <model>` | 모델 정보 확인 |
| `ollama ps` | 현재 실행 중인 모델 확인 |
| `ollama stop <model>` | 모델 실행 종료 |
| `ollama rm <model>` | 모델 삭제 |

```python
from ollama import chat

response = chat(
    model="gemma4:e4b",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "2002년 월드컵에서 한국은 어디까지 진출했어?"},
    ],
)

print(response.message.content)
```

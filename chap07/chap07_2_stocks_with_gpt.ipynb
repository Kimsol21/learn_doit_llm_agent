#### yfinance 란?
- 야후 파이낸스의 금융 데이터를 쉽게 가져올 수 있게 해주는 오픈소스 파이썬 라이브러리.
- 주가, 재무재표, 거래량 등 다양한 데이터를 데이터프레임 형태로 가져올 수 있다.

```python
import yfinance as yf
from rich.pretty import pprint

# Microsoft (MSFT)에 대한 Ticker 객체 생성
msft = yf.Ticker("MSFT")

# Ticker 객체에 대한 정보 출력 (.py에서 실행할 때는 print(msft.info)로 사용)
pprint(msft.info)
```

```python
hist = msft.history(period="5d") # 5일간의 주가 데이터를 가져옴
display(hist) # 데이터 출력
```

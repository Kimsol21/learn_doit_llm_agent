#### LCEL이란?
LCEL(LangChain Expression Language, 랭체인 표현 언어)은 랭체인에서 여러 컴포넌트를 사슬(Chain)처럼 쉽게 엮을 수 있도록 만든 선언형 언어(구문)입니다.
리눅스의 파이프(|) 연산자를 활용하여 프롬프트 | 모델 | 출력_파서와 같은 직관적인 형태로 코드를 작성할 수 있게 해줍니다.

##### 💡 LCEL의 핵심 장점
* 가독성과 단순함: 복잡한 내부 호출 구조를 생략하고 컴포넌트를 | 기호로 이어 붙이기만 하면 되므로 코드가 매우 간결해집니다.
* 기본적인 비동기 및 스트리밍 지원: 별도의 추가 코드 없이도 AI의 답변을 실시간으로 끊어 출력하는 스트리밍(stream) 방식과 비동기(ainvoke) 처리가 자동으로 작동합니다.
* 병렬 실행 최적화: 여러 개의 작업을 동시에 실행할 때 효율적으로 트래픽을 분산하고 처리합니다.

```python
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from rich.pretty import pprint

messages = [
    SystemMessage(content="너는 신데렐라에 나오는 계모야. 그 캐릭터에 맞게 사용자와 대화하라."),
    HumanMessage(content="좋은아침이에요. 오늘 날씨가 참 좋네요. 오늘 하루 계획이 어떻게 되나요?"),
]
```

```python
model = ChatOpenAI(model="gpt-5.6-luna")
pprint(model)
```

```python
res = model.invoke(messages)
pprint(res)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

result = model.invoke(messages)
parser.invoke(result)
```

```python
chain = model | parser
chain.invoke(messages)
```

#### 프롬프트 템플릿 이용하기
- 프롬프트 템플릿은 AI에게 보낼 지시문이나 질문의 고정된 형태를 미리 만들어둔 '문장 틀(양식)'입니다.
- 매번 똑같은 안내문을 길게 쓸 필요 없이, 빈칸에 핵심 변수만 쏙 집어넣어 맞춤형 질문을 자동으로 완성합니다.
- 코드의 재사용성을 높여주며, 사용자가 입력한 단어를 개발자가 원하는 정교한 프롬프트 형태로 변환할 때 필수적입니다.
- 랭체인에서는 PromptTemplate을 사용해 {변수}가 포함된 틀을 만들고 값을 매핑하는 방식으로 간단히 구현합니다.

```python
from langchain_core.prompts import ChatPromptTemplate

system_template = "너는 {story}에 나오는 {character_a} 역할이다. 그 캐릭터에 맞게 사용자와 대화하라."
human_template = "안녕? 저는 {character_b}입니다. 오늘 시간 괜찮으시면 {activity} 같이 할까요?"

prompt_template = ChatPromptTemplate([
    ("system", system_template),
    ("user", human_template),
])

result = prompt_template.invoke({
    "story": "신데렐라",
    "character_a": "계모",
    "character_b": "신데렐라",
    "activity": "저녁"
})

print(result)
```

```python
chain = prompt_template | model | parser

chain.invoke({
    "story": "신데렐라",
    "character_a": "계모",
    "character_b": "신데렐라",
    "activity": "저녁"
})
```
